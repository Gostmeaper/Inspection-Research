```python
"""Setup Notebook"""

from pathlib import Path
import duckdb

# Resolve paths relative to this notebook's location, regardless of cwd
BASE_DIR = Path(__file__).parent if "__file__" in dir() else Path.cwd()

DB_FILE = (BASE_DIR / "../data/processed/Compuspections_merge.parquet").resolve()

print(f"DB_FILE: {DB_FILE}")
print(f"Exists: {DB_FILE.exists()}")

con = duckdb.connect()
```

```python
"""Inspect Columns"""

query = f"""
DESCRIBE SELECT * FROM '{DB_FILE}'
"""

con.execute(query).fetchdf()
```

```python

query = f"""
SELECT *
FROM '{DB_FILE}'
LIMIT 20
"""

con.execute(query).fetchdf()
```

```python

query = f"""
SELECT *
FROM '{DB_FILE}'
USING SAMPLE 20 ROWS
"""

con.execute(query).fetchdf()
```

```python
query = f"""
SELECT COUNT(*)
FROM '{DB_FILE}'
"""

con.execute(query).fetchdf()
```

```python
query = f"""
SELECT State, COUNT(*) AS count
FROM '{DB_FILE}'
GROUP BY State
ORDER BY count DESC
LIMIT 20
"""

con.execute(query).fetchdf()
```
